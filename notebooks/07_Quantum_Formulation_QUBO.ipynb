# Quantum Optimization for Distributed Order Management (DOM)

# Notebook 07 – QUBO Formulation

## Objective

This notebook reformulates the Distributed Order Management (DOM) optimization problem as a **Quadratic Unconstrained Binary Optimization (QUBO)** model.

QUBO is the standard mathematical representation used by many quantum optimization algorithms, including the Quantum Approximate Optimization Algorithm (QAOA). The formulation developed here serves as the bridge between the classical optimization model and the quantum optimization workflow.

```python
import pandas as pd
import numpy as np

print("Libraries Imported Successfully!")
```

# Load Classical Optimization Results

The optimized order assignments generated in the previous notebook are loaded.

These results provide the input required to construct the QUBO representation of the optimization problem.

```python
orders = pd.read_csv("../data/advanced_optimization_results.csv")

print("Dataset Loaded Successfully!")
```

# Prepare QUBO Dataset

A small subset of optimized customer orders is selected for QUBO formulation.

Using a limited number of variables keeps the quantum optimization problem computationally manageable while demonstrating the complete workflow.

```python
qubo_orders = orders.head(10).copy()

print("Quantum Sample Shape:", qubo_orders.shape)
```

# Inspect Optimization Data

The selected customer orders are displayed to verify that the dataset is ready for QUBO formulation.

This verification step helps ensure that all required information is available before constructing the mathematical model.

```python
qubo_orders.head()
```

### Observation

The dataset contains the optimized customer orders that will be converted into binary decision variables for the QUBO model.

# Create Binary Decision Variables

Each customer order is represented using a binary decision variable.

A value of **1** indicates that the corresponding order is selected.

A value of **0** indicates that the order is not selected.

These binary variables form the foundation of the QUBO model.

```python
variables = []

for i in qubo_orders.index:
    variables.append(f"x_{i}")

print("Binary Variables:")
print(variables)
```

# Construct the QUBO Objective Function

The QUBO objective assigns a numerical weight to every binary decision variable.

These weights represent the business value of selecting each customer order.

The objective function will later be optimized using quantum algorithms.

```python
objective = {}

for i in qubo_orders.index:
    objective[f"x_{i}"] = qubo_orders.loc[i, "Order_SKU_Revenue"]

objective
```

# Generate the QUBO Expression

The QUBO objective function is assembled by combining all binary variables and their corresponding coefficients.

This mathematical representation is the input required by quantum optimization algorithms.

```python
for var, value in objective.items():
    print(f"{var} -> {value}")
```

# Display the QUBO Model

The complete QUBO objective function is displayed.

This representation allows verification of the mathematical formulation before it is passed to a quantum optimizer.

```python
print("QUBO Objective:")
print("Maximize Revenue while satisfying constraints")
print("Later this objective will be converted into a QUBO matrix.")
```

### Observation

The generated QUBO model successfully represents the optimization problem using binary variables and weighted coefficients.

This formulation is suitable for quantum optimization techniques such as QAOA.

# Save the QUBO Representation

The generated QUBO coefficients are stored for later use.

The saved formulation will be imported into the Qiskit implementation developed in the next notebook.

```python
objective_df = pd.DataFrame(
    list(objective.items()),
    columns=["Variable", "Revenue"]
)

objective_df.to_csv(
    "../data/qubo_objective.csv",
    index=False
)

print("QUBO Objective Saved Successfully!")
```

# Conclusion

This notebook successfully transformed the Distributed Order Management optimization problem into a Quadratic Unconstrained Binary Optimization (QUBO) formulation.

The generated QUBO model preserves the business objective in a mathematical form that is compatible with quantum optimization algorithms.
