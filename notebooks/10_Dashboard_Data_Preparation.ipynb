# Notebook 10 – Dashboard Data Preparation

## Objective

This notebook prepares the final optimization results for visualization in the Streamlit dashboard.

The processed data provides planners and business users with an easy-to-understand summary of the optimization outcomes and key performance indicators (KPIs).

```python
import pandas as pd

df = pd.read_csv("../data/optimized_orders_v2.csv")
```

# Load Dashboard Dataset

The optimized dataset is loaded and verified before generating dashboard metrics.

This step ensures that the dashboard uses the latest optimization results.

```python
print(df.shape)
print(df.columns)
df.head()
```

### Observation

The dataset has been successfully loaded and is ready for dashboard preparation.

# Generate Dashboard Metrics

Important business metrics are extracted from the optimization results.

These metrics provide planners with a quick overview of optimization performance without examining the complete dataset.

```python
summary = {
    "Total Orders": df["Group_Flag"].nunique(),
    "Total SKUs": df["MaterialNumber"].nunique(),
    "Selected SKUs": df["Selected"].sum(),
    "Not Selected": (df["Selected"] == 0).sum(),
    "Total Revenue": df["Order_SKU_Revenue"].sum(),
    "Average Fill Rate": df["FillRateThreshold"].mean()
}

summary
```

# Create Dashboard Summary Table

The calculated metrics are organized into a structured summary table.

This table serves as the primary data source for the Streamlit dashboard.

```python
summary_df = pd.DataFrame(summary.items(), columns=["Metric", "Value"])
summary_df
```

### Observation

The summary table contains the key business indicators required for dashboard visualization.

# Save Dashboard Data

The dashboard summary is saved as a CSV file.

The Streamlit application reads this file to display business insights and optimization results.

```python
summary_df.to_csv("../data/dashboard_summary.csv", index=False)
```

### Observation

The dashboard dataset has been successfully generated and stored for visualization.

# Conclusion

This notebook prepared the final dashboard dataset by summarizing the optimization results into business-oriented performance metrics.

The generated summary enables the Streamlit dashboard to present optimization insights in a simple, interactive, and decision-friendly format.
