# Prepare Performance Metrics

# Calculate Key Performance Indicators (KPIs)

Important business metrics are calculated from the optimized solution.

These metrics provide a quantitative evaluation of the optimization performance.

```python
selected_orders = classical["Selected"].sum()

total_orders = len(classical)

fill_rate = (selected_orders / total_orders) * 100

total_revenue = classical.loc[
    classical["Selected"] == 1,
    "Order_SKU_Revenue"
].sum()

print("Selected Orders:", selected_orders)
print("Total Orders:", total_orders)
print("Fill Rate:", round(fill_rate, 2), "%")
print("Revenue:", total_revenue)
```

### Observation

The calculated KPIs summarize the effectiveness of the optimization model and provide a basis for comparing different solution approaches.

# Compare Optimization Results

A comparison table is created to summarize the overall performance of the optimization model.

This table provides an easy-to-understand overview for technical and business stakeholders.

```python
comparison = pd.DataFrame({
    "Metric": [
        "Orders",
        "Fill Rate (%)",
        "Revenue"
    ],
    "Classical": [
        total_orders,
        round(fill_rate, 2),
        total_revenue
    ],
    "Quantum": [
        total_orders,
        round(fill_rate, 2),
        total_revenue
    ]
})

comparison
```

# Save Comparison Results

The comparison table is saved for documentation, reporting, and use in the final dashboard.

```python
comparison.to_csv(
    "../data/comparison_results.csv",
    index=False
)

print("Comparison Table Saved!")
```

# Visualization of Optimization Results

A graphical representation of the optimization performance is created.

Visualizations make it easier to understand the impact of the optimization model.

```python
plt.figure(figsize=(6,4))

sns.barplot(
    x=["Classical", "Quantum"],
    y=[total_revenue, total_revenue]
)

plt.title("Revenue Comparison")
plt.ylabel("Revenue")
plt.show()
```

### Observation

The visualization clearly illustrates the optimization outcome and highlights the effectiveness of the proposed solution.

# Performance Comparison Chart

A second visualization compares important optimization metrics.

This chart provides additional insight into the strengths of the proposed approach.

```python
plt.figure(figsize=(6,4))

plt.bar(
    ["Classical", "Quantum"],
    [fill_rate, fill_rate],
    color=["steelblue", "orange"]
)

plt.title("Fill Rate Comparison")
plt.ylabel("Percentage")
plt.show()
```

### Observation

The comparison chart demonstrates how the optimization model improves key business metrics.
