# Quantum Optimization for Distributed Order Management (DOM)

# Notebook 08 – Quantum Optimization using QAOA (Qiskit)

## Objective

This notebook demonstrates how the Distributed Order Management (DOM) optimization problem can be represented using Qiskit.

The QUBO model generated in the previous notebook is converted into a Quantum Optimization model that can be solved using the Quantum Approximate Optimization Algorithm (QAOA).

Since current quantum hardware has limitations, this notebook focuses on preparing and validating the quantum model using simulation.

```python
import pandas as pd
import numpy as np

from qiskit_optimization import QuadraticProgram

print("Libraries Imported Successfully!")
```

# Load the QUBO Model

The QUBO coefficients generated in the previous notebook are loaded.

These coefficients represent the optimization objective in a form suitable for quantum optimization algorithms.

```python
qubo = pd.read_csv("../data/qubo_objective.csv")

print("QUBO Data Loaded Successfully!")
```

# Inspect the QUBO Data

The QUBO dataset is displayed to verify that all variables and coefficients have been generated correctly before constructing the quantum optimization model.

```python
qubo.head()
```

### Observation

The QUBO dataset contains the binary decision variables and their corresponding objective coefficients.

These values will be used to construct the quantum optimization problem.

# Create the Quantum Optimization Model

A Qiskit **QuadraticProgram** object is created.

This object provides the mathematical representation required for quantum optimization algorithms such as QAOA.

```python
qp = QuadraticProgram()

print("Quadratic Program Created!")
```

# Define Binary Decision Variables

Each QUBO variable is added to the Quantum Optimization model as a binary decision variable.

These variables represent whether a customer order is selected or not.

```python
for var in qubo["Variable"]:
    qp.binary_var(name=var)

print("Binary Variables Added!")
```

# Build the Objective Function

The objective function is constructed using the QUBO coefficients.

The quantum optimizer will attempt to maximize the business objective while satisfying the optimization model.

```python
linear = {}

for _, row in qubo.iterrows():
    linear[row["Variable"]] = row["Revenue"]

qp.maximize(linear=linear)

print("Objective Function Added!")
```

# Display the Quantum Optimization Model

The complete Quadratic Program is displayed.

This allows verification that the optimization problem has been correctly converted from the QUBO representation into the Qiskit model.

```python
print(qp)
```

### Observation

The Quadratic Program correctly represents the optimization problem using binary decision variables and the corresponding objective function.

This confirms that the model is ready for quantum optimization.

# Save the Quantum Model

The generated Quantum Optimization model is saved for documentation and future reference.

Saving the model improves reproducibility and simplifies debugging.

```python
with open("../data/qiskit_model.txt", "w") as f:
    f.write(str(qp))

print("Model Saved Successfully!")
```
